```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd
import numpy as np
```

```python

iris = datasets.load_iris()
X = pd.DataFrame(iris.data)
X.columns = ['Sepal_Length', 'Sepal_Width', 'Petal_Length', 'Petal_Width']
print(X)
y = pd.DataFrame(iris.target)
y.columns = ['Targets']
print(y)
```

```python

x_train,x_test,y_train,y_test = train_test_split(iris.data,iris.target,test_size=0.1)
print("Dataset is split into training and testing...")
print("Size of training data and its label",x_train.shape,y_train.shape)
print("Size of testing data and its label",x_test.shape,y_test.shape)
```

```python
from sklearn.svm import SVC
model=SVC()
SVC=model.fit(x_train,y_train)
```

```python
y_pred=SVC.predict(x_test)
y_pred
```

```python
new_data_point = np.array([[6.1, 3.4, 4.6, 1.4]])
predicted_species_id = SVC.predict(new_data_point)
predicted_species_name = iris.target_names[predicted_species_id][0]
```

```python

print("\nPrediction for new data point [6.1, 3.4, 4.6, 1.4]:")
print(f"Predicted Species: {predicted_species_name}")
```

```python

# Show which data points are the support vectors
print(f"\nNumber of support vectors: {len(SVC.support_vectors_)}")
```

```python
print('Confusion Matrix')
print(confusion_matrix(y_test,y_pred))
print('Accuracy Ketrics')
print(classification_report(y_test,y_pred))
```
